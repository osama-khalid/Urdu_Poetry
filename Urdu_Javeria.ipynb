```python
new_text = re.sub(' +',' ',text)
```

```python
clean = []
for item in new_text.split('\n'):
  if len(item)>2:
    clean.append(item)
```

```python
import tqdm
word_count = {}
for sent in tqdm.tqdm(clean):
  predicted_pos_tags = pos_tagger.pos_tags_urdu(sent)
  for word in predicted_pos_tags:
    if word['POS_Tag'][0]=='N':
      if word['Word'] not in word_count:
        word_count[word['Word']] = 0

      word_count[word['Word']] += 1
```

```python
best_words  =sorted(word_count.items(), key=lambda item: item[1])[::-1]
```

```python
import csv

with open('word_count_noun.csv', mode='w') as employee_file:
    employee_writer = csv.writer(employee_file, delimiter=',')

    for word in best_words[:100]:
      employee_writer.writerow([word[0],word[1]])
```
